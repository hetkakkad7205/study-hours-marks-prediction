```python
# Distribution of Scores
# Histogram of final marks
plt.figure(figsize=(8, 5))
sns.histplot(data, x="FinalMarks", bins=20, kde=True)

plt.title("Distribution of Final Marks")
plt.xlabel("Final Marks")
plt.ylabel("Number of Students")
plt.show()

# Histogram of CGPA
plt.figure(figsize=(8, 5))
sns.histplot(data, x="CGPA", bins=15, kde=True)

plt.title("Distribution of CGPA")
plt.xlabel("CGPA")
plt.ylabel("Number of Students")
plt.show()

# Box plot to identify possible outliers
plt.figure(figsize=(8, 4))
sns.boxplot(data, x="FinalMarks")

plt.title("Box Plot of Final Marks")
plt.xlabel("Final Marks")
plt.show()
```

```python
# Correlation HeatMap

# Select numerical columns and exclude student ID
numeric_df = df.select_dtypes(include="number").drop(
    columns=["SID"], errors="ignore"
)

# Calculate correlation matrix
correlation_matrix = numeric_df.corr()

print(correlation_matrix.round(2))

# Plot heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Student Performance")
plt.tight_layout()
plt.show()
```
